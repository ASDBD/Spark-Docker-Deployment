# PySpark Word Count on HDFS

This notebook demonstrates a simple Word Count application using PySpark, reading data from HDFS.

```python
from pyspark.sql import SparkSession

# Initialize SparkSession
spark = SparkSession.builder \
    .appName("HDFS Word Count") \
    .master("spark://spark-master:7077") \
    .getOrCreate()

sc = spark.sparkContext
```

```python
# Read text file from HDFS
hdfs_path = "hdfs://namenode:8020/data/alice_in_wonderland.txt"
text_file = sc.textFile(hdfs_path)

# Show first few lines to verify read
# text_file.take(5)
```

```python
# Perform Word Count
counts = text_file.flatMap(lambda line: line.split(" ")) \
             .map(lambda word: (word, 1)) \
             .reduceByKey(lambda a, b: a + b)

# Collect and print results
output = counts.collect()
#for (word, count) in output:
#    print("%s: %i" % (word, count))
print(output)
```

```python
# Stop SparkSession
spark.stop()
```
